```python
drop_cols = [
    "Customer Email", "Customer Password", "Customer Fname", "Customer Lname",
    "Customer Street", "Latitude", "Longitude",
    "Product Description", "Product Image", "Product Status"
]
df = df.drop(columns=drop_cols)
```

```python
len(df.columns)
```

```python
df["order date (DateOrders)"] = pd.to_datetime(df["order date (DateOrders)"])
df["shipping date (DateOrders)"] = pd.to_datetime(df["shipping date (DateOrders)"])
```

```python
df["shipping_delay_days"] = df["Days for shipping (real)"] - df["Days for shipment (scheduled)"]
df["is_late"] = df["shipping_delay_days"] > 0
df["profit_margin_pct"] = df["Order Profit Per Order"] / df["Sales"] * 100

df["order_month"] = df["order date (DateOrders)"].dt.month
df["order_year"] = df["order date (DateOrders)"].dt.year
df["order_weekday"] = df["order date (DateOrders)"].dt.day_name()
```

```python
def delay_bucket(x):
    if x < 0:
        return "Early"
    elif x == 0:
        return "On-Time"
    elif x <= 2:
        return "1-2 Days Late"
    else:
        return "3+ Days Late"

df["delay_category"] = df["shipping_delay_days"].apply(delay_bucket)
```

```python
df.head()
```

```python
float_cols = df.select_dtypes(include="float").columns
df[float_cols] = df[float_cols].round(2)
```

```python
category_cols = [
    "Type",
    "Delivery Status",
    "Category Name",
    "Customer Country",
    "Customer Segment",
    "Customer State",
    "Department Name",
    "Market",
    "Order Country",
    "Order Region",
    "Order State",
    "Order Status",
    "Shipping Mode",
    "order_weekday",
    "delay_category"
]

for col in category_cols:
    df[col] = df[col].astype("category")
```

```python
int_cols = [
    "Days for shipping (real)",
    "Days for shipment (scheduled)",
    "Category Id",
    "Customer Id",
    "Department Id",
    "Order Customer Id",
    "Order Id",
    "Order Item Cardprod Id",
    "Order Item Id",
    "Order Item Quantity",
    "Product Card Id",
    "Product Category Id",
    "shipping_delay_days",
    "order_month",
    "order_year"
]

for col in int_cols:
    df[col] = df[col].astype("int32")
```

```python
df.info()
```

```python
dim_customer = df[[
    "Customer Id", "Customer Segment", "Customer City",
    "Customer State", "Customer Country", "Customer Zipcode"
]].drop_duplicates(subset="Customer Id").reset_index(drop=True)
```

```python
dim_customer.head()
```

```python
dim_product = df[[
    "Product Card Id", "Product Name", "Product Category Id",
    "Category Name", "Product Price"
]].drop_duplicates(subset="Product Card Id").reset_index(drop=True)
```

```python
print(dim_product["Product Price"].map(repr).head())
```

```python
dim_product.head()
```

```python
dim_department = df[[
    "Department Id", "Department Name"
]].drop_duplicates(subset="Department Id").reset_index(drop=True)
```

```python
dim_department.head()
```

```python
dim_shipping = df[["Shipping Mode"]].drop_duplicates().reset_index(drop=True)

dim_shipping["ShippingModeKey"] = dim_shipping.index + 1

df = df.merge(
    dim_shipping,
    on="Shipping Mode",
    how="left"
)
```

```python
dim_shipping.head()
```

```python
dim_location = df[[
    "Order Country",
    "Order State",
    "Order City",
    "Market",
    "Order Region"
]].drop_duplicates().reset_index(drop=True)

dim_location["LocationKey"] = dim_location.index + 1

df = df.merge(
    dim_location,
    on=[
        "Order Country",
        "Order State",
        "Order City",
        "Market",
        "Order Region"
    ],
    how="left"
)
```

```python
dim_location.head()
```

```python
dim_date = pd.DataFrame({
    "Date": pd.date_range(
        start=df["order date (DateOrders)"].min(),
        end=df["order date (DateOrders)"].max()
    )
})

dim_date["DateKey"] = dim_date["Date"].dt.strftime("%Y%m%d").astype(int)
dim_date["Year"] = dim_date["Date"].dt.year
dim_date["Quarter"] = dim_date["Date"].dt.quarter
dim_date["Month"] = dim_date["Date"].dt.month
dim_date["MonthName"] = dim_date["Date"].dt.month_name()
dim_date["Weekday"] = dim_date["Date"].dt.day_name()
dim_date["WeekdayNumber"] = dim_date["Date"].dt.weekday
dim_date["Weekend"] = dim_date["WeekdayNumber"] >= 5

df["DateKey"] = df["order date (DateOrders)"].dt.strftime("%Y%m%d").astype(int)
```

```python
dim_date.head()
```

```python
fact_orders = df[
[
    "Order Id",
    "Order Item Id",

    "Customer Id",
    "Product Card Id",
    "Department Id",

    "ShippingModeKey",
    "LocationKey",
    "DateKey",

    "shipping date (DateOrders)",

    "Days for shipping (real)",
    "Days for shipment (scheduled)",

    "shipping_delay_days",
    "is_late",
    "delay_category",

    "Late_delivery_risk",
    "Delivery Status",
    "Order Status",

    "Sales",
    "Order Item Total",
    "Order Profit Per Order",

    "Order Item Discount",
    "Order Item Discount Rate",

    "Order Item Quantity",

    "Order Item Product Price",

    "profit_margin_pct"
]
]
```

```python
fact_orders.head()
```

```python
def to_snake(col):
    return (col.lower()
               .replace(" (dateorders)", "")
               .replace(" ", "_"))

fact_orders.columns = [to_snake(c) for c in fact_orders.columns]
dim_customer.columns = [to_snake(c) for c in dim_customer.columns]
dim_product.columns = [to_snake(c) for c in dim_product.columns]
dim_department.columns = [to_snake(c) for c in dim_department.columns]
dim_shipping.columns = [to_snake(c) for c in dim_shipping.columns]
dim_location.columns = [to_snake(c) for c in dim_location.columns]
```

```python
dim_customer.head()
```

```python
fact_orders.to_sql("fact_orders", engine, if_exists="replace", index=False)
dim_customer.to_sql("dim_customer", engine, if_exists="replace", index=False)
dim_product.to_sql("dim_product", engine, if_exists="replace", index=False)
dim_department.to_sql("dim_department", engine, if_exists="replace", index=False)
dim_shipping.to_sql("dim_shipping", engine, if_exists="replace", index=False)
dim_location.to_sql("dim_location", engine, if_exists="replace", index=False)
dim_date.to_sql("dim_date", engine, if_exists="replace", index=False)
```
